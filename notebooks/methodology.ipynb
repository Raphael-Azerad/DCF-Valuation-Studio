# DCF Valuation Studio Methodology

This notebook is a concise reference for the valuation logic used in the Streamlit app. It is not intended to be a full corporate finance textbook; the goal is to document the formulas, assumptions, and limitations behind the model.

## DCF Overview

The app uses a standard enterprise-value discounted cash flow workflow for operating companies:

1. Pull annual financial statements and market data.
2. Normalize revenue, margins, and free cash flow.
3. Forecast free cash flow over an explicit horizon.
4. Estimate cost of equity with CAPM.
5. Estimate WACC from market-value equity and debt.
6. Discount projected free cash flows at WACC.
7. Estimate terminal value with Gordon Growth.
8. Bridge enterprise value to equity value by subtracting debt and adding cash.
9. Divide by shares outstanding to estimate intrinsic value per share.

## CAPM

Cost of Equity = Risk-Free Rate + Beta x Market Risk Premium

CAPM is used as a practical market-based estimate of the equity return investors require for bearing systematic risk.

## WACC

WACC = (E / V x Re) + (D / V x Rd x (1 - Tax Rate))

The model uses market capitalization for equity value and reported debt for debt value. This is a simplification; a production model may use market value of debt, lease adjustments, excess cash adjustments, or more detailed capital-structure assumptions.

## Terminal Value

Terminal Value = Final Forecast FCF x (1 + g) / (WACC - g)

The terminal growth rate must be below WACC. When terminal growth is very close to WACC, the terminal value becomes highly sensitive and can dominate the valuation.

## Example Interpretation

AAPL is a cleaner default example than JPM for this app because Apple is an operating company with interpretable free cash flow and capital structure. The model output should still be read as a valuation exercise based on selected assumptions, not as investment advice or a price target.

## Sensitivity Analysis

DCF outputs are usually most sensitive to WACC, terminal growth, and normalized free-cash-flow margin. The app includes a two-way WACC / terminal-growth matrix and Bear/Base/Bull scenario analysis to make that sensitivity visible.

## Limitations

- Public market data can be delayed, restated, or missing.
- Financial institutions require special care because debt, cash, interest expense, working capital, and regulatory capital do not behave like they do for industrial companies.
- The model does not include stock-based compensation adjustments, lease capitalization, segment forecasts, or share-count forecasts.
- Output depends heavily on selected assumptions and should not be treated as investment advice.
